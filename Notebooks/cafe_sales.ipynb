```python
df.fillna({'Item':"unknown"}, inplace = True)
df.fillna({'Location':"unknown"}, inplace = True)
df.fillna({'Payment Method':"unknown"}, inplace = True)
df.fillna({'Quantity':0}, inplace = True)
df.fillna({'Price Per Unit':0}, inplace = True)
df.fillna({'Total Spent':0}, inplace = True)
df.isnull().sum()
```

```python
df['Transaction Date'].isin(['UNKNOWN']).value_counts()

df['Transaction Date'] = df['Transaction Date'].replace(['UNKNOWN', 'ERROR'], '1900-01-01')
df['Transaction Date'].isin(['UNKNOWN']).value_counts()

df['Transaction Date'] = pd.to_datetime(df['Transaction Date'])

df.fillna({'Transaction Date': '1900-01-01'}, inplace= True)
print(df.isnull().sum())

print(df.sample(10))
```

```python
df.info()
```

```python


print(f"{df['Item'].isin(['UNKNOWN', 'ERROR']).value_counts()}\n")
df['Item'] = df['Item'].replace(['UNKNOWN', 'ERROR'], 'unknown')
print(f"{df['Item'].isin(['UNKNOWN', 'ERROR']).value_counts()}\n")

print(f"{df['Location'].isin(['UNKNOWN', 'ERROR']).value_counts()}\n")
df['Location'] = df['Location'].replace(['UNKNOWN', 'ERROR'], 'unknown')
print(f"{df['Location'].isin(['UNKNOWN', 'ERROR']).value_counts()}\n")

print(f"{df['Payment Method'].isin(['UNKNOWN', 'ERROR']).value_counts()}\n")
df['Payment Method'] = df['Payment Method'].replace(['UNKNOWN', 'ERROR'], 'unknown')
print(f"{df['Payment Method'].isin(['UNKNOWN', 'ERROR']).value_counts()}\n")

print(f"{df['Quantity'].isin(['UNKNOWN', 'ERROR']).value_counts()}\n")
df['Quantity'] = df['Quantity'].replace(['UNKNOWN', 'ERROR'], 'unknown')
print(f"{df['Quantity'].isin(['UNKNOWN', 'ERROR']).value_counts()}\n")


print(f"{df['Price Per Unit'].isin(['UNKNOWN', 'ERROR']).value_counts()}\n")
df['Price Per Unit'] = df['Price Per Unit'].replace(['UNKNOWN', 'ERROR'], 'unknown')
print(f"{df['Price Per Unit'].isin(['UNKNOWN', 'ERROR']).value_counts()}\n")


print(f"{df['Total Spent'].isin(['UNKNOWN', 'ERROR']).value_counts()}\n")
df['Total Spent'] = df['Total Spent'].replace(['UNKNOWN', 'ERROR'], 'unknown')
print(f"{df['Total Spent'].isin(['UNKNOWN', 'ERROR']).value_counts()}\n")
```

```python
df.to_csv('cleaned_cafe_sales.csv', index=False)
```

```python
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
df['Price Per Unit'] = pd.to_numeric(df['Price Per Unit'], errors='coerce')
df['Total Spent'] = pd.to_numeric(df['Total Spent'], errors='coerce')
```

```python
df.loc[df['Quantity'].isna(), 'Quantity'] = df['Total Spent'] / df['Price Per Unit']
df.loc[df['Price Per Unit'].isna(), 'Price Per Unit'] = df['Total Spent'] / df['Quantity']
df.loc[df['Total Spent'].isna(), 'Total Spent'] = df['Quantity'] * df['Price Per Unit']
```
